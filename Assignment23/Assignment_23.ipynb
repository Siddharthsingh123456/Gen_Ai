## RAG concept

Retrieval-Augmented Generation first retrieves relevant external documents and then gives those documents to an LLM as context. This can improve factual grounding and allows an application to answer questions about information outside the model's original training data.

**Pipeline:** documents → chunks → embeddings → vector store → retrieval → context → LLM → answer.

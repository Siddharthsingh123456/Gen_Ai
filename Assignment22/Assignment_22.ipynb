## Key concepts

**Embeddings:** numerical vectors representing semantic meaning.

**Cosine similarity:** compares the direction of two vectors; higher similarity usually means the texts are more semantically related.

**FAISS:** a fast local similarity-search/indexing library.

**ChromaDB:** a persistent vector-store/database with collections.

**RAG:** retrieve relevant chunks using vector similarity before sending context to a language model.

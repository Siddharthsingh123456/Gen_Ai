## Task 3 — Groq-Based RAG Pipeline

The retrieval stage finds relevant chunks. The prompt then passes only the retrieved
context to the Groq model. The included example uses FAISS with OpenAI embeddings.


```python
try:
    retriever, format_docs, prompt, llm, chunks = __import__("main").build_rag_chain()
    question = "Why should API keys be stored in environment variables?"
    docs = retriever.invoke(question)
    context = format_docs(docs)
    answer = llm.invoke(prompt.invoke({"context": context, "question": question}))

    print("Number of chunks:", len(chunks))
    print("\nRetrieved context:\n", context)
    print("\nAnswer:\n", answer.content)
except Exception as e:
    print(type(e).__name__, e)
```

## Task 4 — Prompt Template for Groq RAG

```python
from langchain_core.prompts import ChatPromptTemplate

rag_prompt = ChatPromptTemplate.from_messages([
    ("system", "Answer only from the retrieved context."),
    ("human", "Context:\n{context}\n\nQuestion:\n{question}")
])

print(rag_prompt.format(
    context="FastAPI is a Python API framework.",
    question="What is FastAPI?"
))
```
